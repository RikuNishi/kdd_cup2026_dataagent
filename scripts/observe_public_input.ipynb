```python
difficulty_tasks = {}

for task in input_summary['tasks']:
    difficulty = task['task_json']['difficulty']
    if difficulty not in difficulty_tasks:
        difficulty_tasks[difficulty] = []
    difficulty_tasks[difficulty].append(task['task_id'].replace('task_', ''))

order = ['easy', 'medium', 'hard', 'extreme']
for difficulty in order:
    task_ids = difficulty_tasks.get(difficulty, [])
    print(f"{difficulty} ({len(task_ids)}件): {task_ids}")
```

```python
order = ['easy', 'medium', 'hard', 'extreme']

difficulty_questions = {}

for task in input_summary['tasks']:
    difficulty = task['task_json']['difficulty']
    if difficulty not in difficulty_questions:
        difficulty_questions[difficulty] = []
    difficulty_questions[difficulty].append((task['task_id'].replace('task_', ''), task['task_json']['question']))

for difficulty in order:
    questions = difficulty_questions.get(difficulty, [])
    print(f"\n=== {difficulty} ({len(questions)}件) ===")
    for task_id, question in questions:
        print(f"  {task_id}: {question}")
```
