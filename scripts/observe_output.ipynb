```python
failed_tasks = [t for t in output_summary['tasks'] if t['succeeded'] is False]

for t in failed_tasks:
    print(t['task_id'], t['failure_reason'])

print('\n総数：', len(failed_tasks))
```

```python
reason_counts = {}
for t in failed_tasks:
    reason = t['failure_reason']
    if reason:
        reason_counts[reason] = reason_counts.get(reason, 0) + 1

for reason, count in sorted(reason_counts.items(), key=lambda x: -x[1]):
    print(f"{count}件: {reason}")
```
