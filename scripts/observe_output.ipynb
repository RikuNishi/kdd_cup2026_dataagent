# Error

```python
with open(f'/Users/temmasaito/Work/IT/Competition/KDDCup2026/kdd_cup2026_dataagent/artifacts/runs/{job_id}/summary.json', 'r', encoding='utf-8') as f:
    output_summary = json.load(f)
```

```python
output_summary
```

```python
failed_tasks = [t for t in output_summary['tasks'] if t['succeeded'] is False]

for t in failed_tasks:
    print(t['task_id'], t['failure_reason'])

print('\n総数：', len(failed_tasks))
```

```python
reason_counts = {}
for t in failed_tasks:
    reason = t['failure_reason']
    if reason:
        reason_counts[reason] = reason_counts.get(reason, 0) + 1

for reason, count in sorted(reason_counts.items(), key=lambda x: -x[1]):
    print(f"{count}件: {reason}")
```

```python
with open(f'/Users/temmasaito/Work/IT/Competition/KDDCup2026/kdd_cup2026_dataagent/artifacts/runs/{job_id}/evaluation.json', 'r', encoding='utf-8') as f:
    output_evaluation = json.load(f)
```

```python
output_evaluation
```

```python
# missing_prediction=False かつ score != 1.0
low_score_tasks = [t for t in output_evaluation['tasks'] if not t['missing_prediction'] and t['score'] != 1.0]

# missing_prediction=False かつ recall != 1.0
low_recall_tasks = [t for t in output_evaluation['tasks'] if not t['missing_prediction'] and t['recall'] != 1.0]
```

```python
print("=== Score が1でないタスク ===")
for t in low_score_tasks:
    print(f"{t['task_id']}: score={t['score']}, recall={t['recall']}, matched={t['matched_columns']}, gold={t['gold_columns']}, predicted={t['predicted_columns']}, extra={t['extra_columns']}")
```

```python
print(f"\n=== Recall が1でないタスク ===")
for t in low_recall_tasks:
    print(f"{t['task_id']}: score={t['score']}, recall={t['recall']}, matched={t['matched_columns']}, gold={t['gold_columns']}, predicted={t['predicted_columns']}, extra={t['extra_columns']}")
```
